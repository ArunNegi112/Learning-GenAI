## What are AI Agents
An ai agents is an intelligent system that recieves a high level goal from user, and based on that it autonomously plans, decides, and executes a sequence of actions with the help of external tools, APIs, or knowledge source   
All this while maintaining the original context and reasoning over multiple steps


AI Agent = loop(LLM + Tools)

Characteristics:
- Goal Driven
- Autonomous planning
- Tool-using
- Content aware and adaptive (Rethinking)

Steps to create an AI Agent:
- Create Tools
- Decide LLM which can work with tool or multiple tools
- Create Agent with appropriate design
- Wrap the agent AgentExecutor
- Use the agent

#### 1. Design patterns in Agents
Choose a Design pattern for your agent like:  



1. ReAct : Reasoning + Action agent  
A ReAct agent is fundamentally a LLM in a loop where it reasons about the current state and users query --> optionally use available tools --> observes the result from tools --> and repeats until it can completely answer the original user query and generate final response

Here we do three things 
- Reasoning and Thinking
- Tool calling and Action
- Observation  
IN A LOOP UNTIL WE CAN ANSWER THE ORIGINAL QUERY

#### 2. Create an Agent
Use appropiate prompt for your required design of Agent.   
Prompt can be written by us or taken from [https://smith.langchain.com/hub/](https://smith.langchain.com/hub/)

#### 3. Agent Executor
Agent executor orchestrates this entire loop.
1. Sends the input and previous messages to agent
2. Gets next `action` from agent
3. Executes the tools for required action 
4. Gets results and add them back to history (Thought Trace)
5. Loops this until agent says "Final Answer"
